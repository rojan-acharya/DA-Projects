## Calculate the average fare price for each vehicle_type.

```python
df.groupby('vehicle_type')['fare_price'].mean().sort_values(ascending=False)
```

```python
df.groupby('customer_cancel_reason')['customer_cancel'].count().sort_values(ascending=False)
```

## Find the average ride distance and average fare price for completed rides.

```python
mask = df['status']=='Completed'

avg_ride = df.loc[mask, 'ride_distance'].mean()
avg_price = df.loc[mask, 'fare_price'].mean()
```

```python
avg_ride, avg_price
```

## Get the total number of rides for each date.

```python
df.groupby('date')['booking_id'].count()
```

## Find the busiest hour of the day for bookings.

```python
# merging date and time column for full time access

df['full_time'] = df['date'] +' '+ df['time']
```

```python
# converting to datetime 

df['full_time'] = pd.to_datetime(df['full_time'])
```

```python
# accessing hour and bookings

hours = df['full_time'].dt.hour
booking_count = df.groupby(hours)['booking_id'].count().sort_values(ascending=False).head(1)
booking_count
```

## Calculate the number of rides on weekends vs weekdays.

```python
def classify_days(day):
    return 'Weekend' if day >= 5 else 'Weekday'

df['day_type'] = df['full_time'].dt.weekday.apply(classify_days)
```

```python
df.groupby('day_type')['booking_id'].count()
```

## Find the top 3 pickup locations with the highest number of bookings.

```python
df.groupby('pickup_location')['booking_id'].count().sort_values(ascending=False).head(3)
```

## For each driver rating, calculate the average fare price and ride distance.

```python
df.groupby('driver_rating').agg(
    avg_price = ('fare_price', lambda x: round(x.mean(), 2)),
    avg_distance = ('ride_distance', lambda x: round(x.mean(), 2))
).reset_index()
```

## Get the percentage of incomplete rides compared to total rides.

```python
(df['incomplete_rides'].sum() / len(df)) * 100
```

## Find the vehicle type with the highest average customer rating.

```python
df.groupby('vehicle_type')['customer_rating'].mean().sort_values(ascending=False).head(1)
```

## Group by vehicle_type and get average fare price, average distance, and average driver rating.

```python
df.groupby('vehicle_type').agg(
    avg_price = ('fare_price', lambda x: x.mean()),
    avg_distance = ('ride_distance', lambda x: x.mean()),
    avg_driver_rating = ('driver_rating', lambda x: x.mean())
).reset_index()
```

## Group by customer_cancel_reason and count cancellations.

```python
df.groupby('customer_cancel_reason').agg(
    total_cancel = ('customer_cancel', lambda x: x.count())
).reset_index()
```
