```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/processed/uber.csv')
```

```python
pd.set_option('display.max_column', None)
```

## What is the average fare price per ride?

```python
df['fare_price'].mean().round(2)
```

## How does fare price vary by vehicle type?

```python
result = df.groupby('vehicle_type')['fare_price'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,6))
df.boxplot(column='fare_price', by='vehicle_type')
plt.xticks(rotation=45)

plt.show()
```

## What is the relationship between ride distance and fare price?

```python
df[['ride_distance','fare_price']].corr()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df['ride_distance'], df['fare_price'], alpha=0.6)
plt.xlabel("Ride distance km.")
plt.ylabel("Fare price")
plt.title("Relationship between Ride Distance and Fare Price")
plt.show()
```
